```python
import pybaseball as pb 
import pandas as pd

SEASON = 2025
WINDOWS = [5, 10, 20]

all_teams = ['ARI','ATL','BAL','BOS','CHC','CHW','CIN','CLE','COL','DET',
             'MIA','HOU','KCR','LAA','LAD','MIL','MIN','NYM','NYY','ATH',
             'PHI','PIT','SDP','SFG','SEA','STL','TBR','TEX','TOR','WSN']
```

```python
# 1. PULL — one schedule_and_record per team, collect into one frame

team_dfs = []
failed = []
for team in all_teams:
    try:
        team_dfs.append(pb.schedule_and_record(SEASON, team=team))
    except Exception as e:
        failed.append((team, str(e)))
        print(f"FAILED: {team} -> {e}")

big_df = pd.concat(team_dfs, ignore_index=True)
```

```python
# 2. CLEAN

big_df['Date'] = big_df['Date'].str.replace(r'\s*\(\d\)', '', regex=True)
big_df['Date'] = pd.to_datetime(big_df['Date'] + f' {SEASON}', format='%A, %b %d %Y')

# Keep only clean 9-inning games (drops sub-9 corrupted + extra-inning games)
big_df = big_df[big_df['Inn'] == 9.0].copy()

# Total runs (target)
big_df['total_runs'] = big_df['R'] + big_df['RA']
```

```python
# 3. FEATURES — rolling averages

big_df = big_df.sort_values(['Tm', 'Date']).reset_index(drop=True)

for w in WINDOWS:
    big_df[f'off_roll{w}'] = (
        big_df.groupby('Tm')['R'].transform(lambda x: x.shift(1).rolling(w).mean())
    )
    big_df[f'def_roll{w}'] = (
        big_df.groupby('Tm')['RA'].transform(lambda x: x.shift(1).rolling(w).mean())
    )

roll_cols = [f'{s}_roll{w}' for w in WINDOWS for s in ('off', 'def')]
```

```python
# 4. BUILD GAME ROWS — home rows = one row per unique game

lookup = big_df[['Tm', 'Date'] + roll_cols].copy()

home = big_df[big_df['Home_Away'] == 'Home'].copy()
home = home.rename(columns={c: f'home_{c}' for c in roll_cols})
away = lookup.rename(columns={c: f'away_{c}' for c in roll_cols})

games = home.merge(
    away,
    left_on=['Opp', 'Date'],
    right_on=['Tm', 'Date'],
    suffixes=('', '_awaydup')
)
```

```python
# 5. TRIM — keep only what the model needs

keep_cols = (
    ['Date', 'Tm', 'Opp', 'total_runs', 'R', 'RA']
    + [f'home_{c}' for c in roll_cols]
    + [f'away_{c}' for c in roll_cols]
)
games = games[keep_cols].rename(columns={'Tm': 'home_team', 'Opp': 'away_team',
                                          'R': 'home_runs', 'RA': 'away_runs'})

# Drop cold-start rows (any NaN rolling value from unfilled windows)
games = games.dropna().reset_index(drop=True)

print(games.shape)
games.head()
```

```python
games.to_csv('games_2025_features.csv', index=False)
print(f"Saved {games.shape[0]} games to games_2025_features.csv")
```
